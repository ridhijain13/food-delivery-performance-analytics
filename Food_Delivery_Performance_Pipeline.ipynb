```python
from pyspark.sql.functions import col, when, to_timestamp, from_unixtime

# Load Bronze data
bronze_df = spark.table("workspace.default.bronze_waybill")

# Convert zero event timestamps to NULL
silver_df = bronze_df

time_columns = [
    "dispatch_time",
    "grab_time",
    "fetch_time",
    "arrive_time"
]

for c in time_columns:
    silver_df = silver_df.withColumn(
        c,
        when(col(c) == 0, None).otherwise(col(c))
    )

# Convert Unix timestamps into readable timestamps
timestamp_columns = [
    "dispatch_time",
    "grab_time",
    "fetch_time",
    "arrive_time",
    "estimate_arrived_time"
]

for c in timestamp_columns:
    silver_df = silver_df.withColumn(
        c + "_readable",
        when(col(c) > 0, to_timestamp(from_unixtime(col(c))))
    )

# Save the Silver table
silver_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.silver_waybill")

print("Silver table created successfully!")
```

```python
from pyspark.sql.functions import col, round, when

# Step 1: Load the Silver table
silver_df = spark.table("workspace.default.silver_waybill")

# Step 2: Calculate delivery duration in minutes
silver_df = silver_df.withColumn(
    "delivery_duration_min",
    when(
        (col("dispatch_time") > 0) & (col("arrive_time") > 0),
        round((col("arrive_time") - col("dispatch_time")) / 60, 2)
    )
)

# Step 3: Calculate delay compared with estimated arrival
silver_df = silver_df.withColumn(
    "delivery_delay_min",
    when(
        (col("arrive_time") > 0) & (col("estimate_arrived_time") > 0),
        round((col("arrive_time") - col("estimate_arrived_time")) / 60, 2)
    )
)

# Step 4: Classify delivery status
silver_df = silver_df.withColumn(
    "delivery_status",
    when(col("arrive_time").isNull(), "Unknown")
    .when(
        col("arrive_time") > col("estimate_arrived_time"),
        "Late"
    )
    .otherwise("On Time")
)

# Step 5: Save the updated Silver table
silver_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.silver_waybill")

print("Delivery metrics added successfully!")

# Step 6: Display the results
display(
    silver_df.select(
        "order_id",
        "delivery_duration_min",
        "delivery_delay_min",
        "delivery_status"
    ).limit(10)
)
```

```python
from pyspark.sql.functions import col, count, sum, avg, round, when

# Load the Silver table
silver_df = spark.table("workspace.default.silver_waybill")

# Calculate overall delivery performance
gold_df = silver_df.agg(
    count("*").alias("total_orders"),

    sum(
        when(col("delivery_status") == "Late", 1).otherwise(0)
    ).alias("late_orders"),

    sum(
        when(col("delivery_status") == "On Time", 1).otherwise(0)
    ).alias("on_time_orders"),

    sum(
        when(col("delivery_status") == "Unknown", 1).otherwise(0)
    ).alias("unknown_orders"),

    round(avg("delivery_duration_min"), 2).alias("avg_delivery_duration_min"),

    round(avg("delivery_delay_min"), 2).alias("avg_delivery_delay_min")
)

# Calculate late delivery percentage using only orders with known status
gold_df = gold_df.withColumn(
    "late_delivery_pct",
    round(
        col("late_orders") * 100 /
        when(
            (col("late_orders") + col("on_time_orders")) > 0,
            col("late_orders") + col("on_time_orders")
        ),
        2
    )
)

# Save the Gold table
gold_df.write.format("delta") \
    .option("mergeSchema", "true") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.gold_delivery_performance")

# Display the results
display(gold_df)

print("Gold delivery performance table created!")
```

```python
%sql
SELECT
    delivery_status,
    COUNT(*) AS total_orders,
    ROUND(
        COUNT(*) * 100.0 /
        (SELECT COUNT(*)
         FROM workspace.default.silver_waybill),
        2
    ) AS percentage
FROM workspace.default.silver_waybill
GROUP BY delivery_status
ORDER BY total_orders DESC;
```

```python
%sql
SELECT
    dt AS delivery_date,
    COUNT(*) AS total_orders,
    SUM(CASE WHEN delivery_status = 'Late' THEN 1 ELSE 0 END) AS late_orders,
    ROUND(
        SUM(CASE WHEN delivery_status = 'Late' THEN 1 ELSE 0 END) * 100.0 /
        SUM(CASE WHEN delivery_status IN ('Late', 'On Time') THEN 1 ELSE 0 END),
        2
    ) AS late_delivery_pct
FROM workspace.default.silver_waybill
GROUP BY dt
ORDER BY dt;
```

```python
%sql
SELECT
    HOUR(TO_TIMESTAMP(dispatch_time)) AS dispatch_hour,
    COUNT(*) AS total_orders,
    ROUND(AVG(delivery_duration_min), 2)
        AS avg_delivery_duration_min,
    ROUND(
        SUM(CASE WHEN delivery_status = 'Late' THEN 1 ELSE 0 END)
        * 100.0 /
        SUM(CASE
            WHEN delivery_status IN ('Late', 'On Time') THEN 1
            ELSE 0
        END),
        2
    ) AS late_delivery_pct
FROM workspace.default.silver_waybill
WHERE dispatch_time IS NOT NULL
  AND dispatch_time > 0
GROUP BY HOUR(TO_TIMESTAMP(dispatch_time))
ORDER BY dispatch_hour;
```

```python
from pyspark.sql.functions import col, hour, to_timestamp, count, avg, sum, when, round

silver_df = spark.table("workspace.default.silver_waybill")

hourly_gold_df = (
    silver_df
    .filter(col("dispatch_time") > 0)
    .withColumn(
        "dispatch_hour",
        hour(to_timestamp(col("dispatch_time")))
    )
    .groupBy("dispatch_hour")
    .agg(
        count("*").alias("total_orders"),
        round(avg("delivery_duration_min"), 2)
            .alias("avg_delivery_duration_min"),
        sum(when(col("delivery_status") == "Late", 1).otherwise(0))
            .alias("late_orders"),
        round(
            sum(when(col("delivery_status") == "Late", 1).otherwise(0)) * 100.0 /
            sum(when(col("delivery_status").isin("Late", "On Time"), 1).otherwise(0)),
            2
        ).alias("late_delivery_pct")
    )
)

hourly_gold_df.write.format("delta").mode("overwrite").saveAsTable(
    "workspace.default.gold_hourly_performance"
)

display(hourly_gold_df.orderBy("dispatch_hour"))
```
